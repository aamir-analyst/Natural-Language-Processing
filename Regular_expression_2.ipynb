## Additional REGEX SYNTAX

### OR Operator

```python
re.search(r"man|woman","This man was here.")
```

```python
re.search(r"man|woman","This woman was here.")
```

## The Wildcard Character

```python
re.findall(r".at","The cat in the hat sat here.")
```

```python
re.findall(r".at","The bat went splat")
```

```python
re.findall(r"...at","The bat went splat.")
```

```python
# One or more non-whitespace that ends with 'at
re.findall(r'\S+at','The bat went splat')
```

```python
# with whitespace
re.findall(r'\s+at','The bat went splat')
```

## Starts With and Ends With

```python
# Ends with a number
re.findall(r"\d$","This ends with a number 2")
```

```python
# Starts with a number
re.findall(r"^\d","1 is the loneliest number")
```

## Exclusion

- To exclude characters, we can use the ^ symble in conjunction with a set of brackets []. Anything inside the brackets is excluded.

```python
pharase = "there are 3 numbers 34 inside 5 this sentence."
```

```python
re.findall(r"[^\d]", pharase)
```

```python
re.findall(r"[^\d]+",pharase)
```

```python
test_pharase = "This is a string But it has punctuation. How can we remove it?"
```

```python
re.findall(r"[^!.? ]+", test_pharase)
```

```python
clean = " ".join(re.findall("[^!.? ]+", test_pharase))
```

```python
clean
```

## Perentheses for Multiple Options

```python
# Find words that start with cat and end with one these options: 'fish', 'nap', or 'claw'
text = 'Hello, world you like some catfish?'
texttwo = 'Hello, world you like to take a catnap?'
textthree = 'Hello, have you seen this caterpillar'
```

```python
re.search(r'cat(fish|nap|claw)', text)
```

```python
re.search(r"cat(fish|nap|claw)", texttwo)
```

```python
re.search(r"cat(fish|nap|claw)", textthree)
```
