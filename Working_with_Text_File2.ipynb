```python
  # 1. Creating and Writing to a Text File

# Open file in write mode
file = open("students.txt", 'w')

file.write("vikas\n")
file.write("Amit\n")
file.write("Priya\n")
file.write("Hena\n")
```

```python
file.close()
```

```python
print("File created successfully")
```

```python
# 2. Write Multiple Lines Using writelines()
students = ["Rahul\n","Priya\n","Amit\n","Neha\n"]
```

```python
with open("students.txt","w") as file:
  file.writelines(students)
```

```python
print("Data written successfully")
```

```python
# 3. Reading the Entire File
with open("students.txt", "r") as file:
  data = file.read()
print(data)
```

```python
# 4. Reading FIle Line by Line
with open("students.txt", "r") as file:
  for line in file:
    print(line.strip()) # print(line)
```

- strip() removes the newline character (\n) from each line

```python
# 5. Using readline()
with open("students.txt", "r") as file:
  line1 = file.readline()
  line2 = file.readline()

print("First line : ", line1.strip())
print("Second line :", line2.strip())
```

```python
# 6. Using readlines
with open("students.txt", "r") as file:
  lines = file.readlines()
print(lines)
```

```python
# You can also process the list:
with open("students.txt", "r") as file:
  lines = file.readlines()
for line in lines:
  print(line.strip())
```

```python
# 7. Appending Data to a file
with open("students.txt", "a") as file:
  file.write("Karan\n")
  file.write("Simran\n")
print("New students added")
```

```python
name = input("Enter your name : ")
age = input("Enter your age : ")

with open("user.txt", "w+") as file:
  file.write("Name : " + name + "\n")
  file.write("Age : " + age + "\n")
print("Information saved")
```

```python
# 9. Counting number of Lines
with open("students.txt", "r") as file:
  lines = file.readlines()
print("Number of lines", len(lines))
```

```python
# 10. Counting Words in a Text File
with open("students.txt", "r") as file:
  data = file.read()

words = data.split()
print("Number of words : ", len(words))
```

```python
# 11. Counting Characters
with open("students.txt", "r") as file:
  data = file.read()
print("Number of characters : ", len(data))
```

```python
# To exclude spaces:
characters = data.replace(" "," ").replace("\n", "")
print("Characters excluding spaces :", len(characters))
```

```python
# 12. Searcheing for a word
search_word = input("Enter the word to search : ")
with open("students.txt", "r") as file:
  data = file.read()
if search_word.lower() in data.lower():
  print("Word Found.")
else:
  print("Word not found")
```

```python
# 13. Counting Occurrences of a Word

search_word = input("Enter word : ")
with open("students.txt", "r") as file:
  data = file.read()
count = data.lower().count(search_word.lower())
print("Occurrences : ", count)
```

```python
# 14. Copying Contents from One File to Another
with open("students.txt", "r") as source:
  data = source.read()

with open("backup.txt", "w") as destination:
  destination.write(data)
print("File copied successfully")
```

```python
# 15. Reading only the First 10 Characters
with open("students.txt", "r") as file:
  data = file.read(10)
print(data)
```
