### Stemming in NLP
**Stemming** is the process of reducing words to their root/base form (stem) by removing prefixes or suffixes.

- **Important:** spacy does not provide a built-in stemming algorithm.
spacy primarily uses lemmatization. If you specifically want stemming while using spacy, you can integrate a stemmer such as NLTk's PorterStemmer or SnowballStemmer.

## Stemming using NLTK

```python
# Install the NLTK library for stemming operations
!pip install nltk
```

```python
# Import the PorterStemmer class from NLTK
from nltk.stem import PorterStemmer

# Initialize the PorterStemmer object for word stemming
stemmer = PorterStemmer()
```

### 1. Basic Stemming

```python
# Define a new list of words sharing a similar root
words = ["computing", "computed", "computes", "computer"]

for word in words:
  print(f"{word:12} -> {stemmer.stem(word)}")
```

### 2. Stemming a Sentence

```python
# Define the sentence
text = "The engineers are designing innovative software solutions."

# Split the sentence into individual words
words = text.split()

# Apply stemming to each word in the sentence
print("--- Stemming Sentence Words ---")
for word in words:
  print(f"{word:12} -> {stemmer.stem(word)}")
word_list = ["designing", "designed", "designs", "designer"]

# Apply stemming to the list of words
print("\n--- Stemming Word List ---")
for word in word_list:
  print(f"{word:12} -> {stemmer.stem(word)}")
```

```python
# Words
words = ["happy", "happier", "happest", "happiness"]

for word in words:
  stem = stemmer.stem(word)
  print(f"{word:15} -> {stemmer.stem(word)}")
```

### 8. Stemming words related to organize

### 9. Stemming text using regular expressions

```python
# Import Regular Expression Library
import re

text = "Data scientists are extracting patterns from massive unstructured datasets."

# Use regex to extract only alphabetic words and convert text to lowercase
words = re.findall(r"\b[a-zA-Z]+\b", text.lower())

for word in words:
  print(f"{word:15} -> {stemmer.stem(word)}")
```

### 10. Create a stemmed sentence

```python
# Define sentence
text = "The developers are building scalable applications and optimizing them continuously."
words = text.lower().split()

# Initialize an empty list to store the stemmed words
stemmed_words = []

for word in words:

  # Stem the word and append it to our new list
  stemmed_words.append(stemmer.stem(word))
stemmed_text = " ".join(stemmed_words)

print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```

### Import the libraries

```python
# Import spaCy for tokenize and NLTK for stemming
import spacy
from nltk.stem import PorterStemmer

# Load the small English model
nlp = spacy.load("en_core_web_sm")

# Initialize the NLTK PorterStemmer object
stemmer = PorterStemmer()
```

### 1. Stemming spaCy tokens

```python
# Define a Sentence
text = "The kids are jumping and running fast."

# Process the text using spaCy
doc = nlp(text)
for token in doc:

  # Check if the token is an alphabetic word
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

- Hare, spaCy performs tokenization, while NLTK performs stemming.

### 2. Stemming different forms of a verb

```python
# Define Sentence
text = "I cook food. She cooks well. They cooked yesterday."

# Process the text to extract tokens
doc = nlp(text)

# Loop through each token
for token in doc:

  # Only process alphabetical words
  if token.is_alpha:

    # Print original token and its stemmed version
    print(f"{token.text:10} -> {stemmer.stem(token.text)}")
```

### 3. Reading-related words

```python
# Define a simple sentence with variations of the word "help"
text = "Helping helpers help helpful friends."
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

### 4. Learning-related words

```python
# Define sentence
text = "Cleaners clean cleaned rooms by cleaning"
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

### 5. Connect-related words

### 7. Remove punctuation before stemming

```python
# Define sentence
text = "Tigers are big, strong, and fast!"
doc = nlp(text)

for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {stemmer.stem(token.text)}")
```

### 8. Stemming with lowercase conversion

```python
# Sentence
text = "Work Working WORKERS Worked"

# Process the text to create tokens
doc = nlp(text)

# Loop through each token in the document
for token in doc:

  # Check if the token is alphabetic
  if token.is_alpha:

    # Convert the token text to lowercase before stemming
    word = token.text.lower()

    # Print the lowercase word and its stemmed version
    print(f"{word:15} -> {stemmer.stem(word)}")
```

### 9. Create a stemmed sentence using spaCy

```python
# Define sentence for testing
text = "The dogs are chasing cats and barking loudly."

doc = nlp(text)

# initialize an empty list to store the stemmed words
stemmed_words = []

for token in doc:
  if token.is_alpha:

    # Convert to lowercase, stem the word, and append to the list
    stemmed_words.append(stemmer.stem(token.text.lower()))

# Join the stemmed words back together into a single string separated by spaces
stemmed_text = " ".join(stemmed_words)

print("Original:")
print(text)

print("\nStemmed:")
print(stemmed_text)
```
