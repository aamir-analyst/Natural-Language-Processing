#### **`Stopwords`** are common words that often carry little useful information for certain NLP tasks. Examples include: the, is, am, are, a, an, and, or, in, on, of, to, for

```python
# Import Library
import spacy

# Load small English NLP model
nlp = spacy.load("en_core_web_sm")
```

### 1: Check Whether a Word is a Stopword

```python
# Load small English model
nlp = spacy.load("en_core_web_sm")

# list of words
words = ["the", "data", "is", "analyzed", "by"]

# Loop through each word
for word in words:
  doc = nlp(word)  # Process the word

  # Check and print if the token is a stopword
  for token in doc:
    print(f"{token.text:15} -> {token.is_stop}")
```

### 2: Identify Stopwords in a Sentence


```python
# Define a Sentence
text = "Python is highly effective for large-scale data analysis."

# Process the text
doc = nlp(text)

# Check and print stopword status for each token
for token in doc:
  print(f"{token.text:10} -> {token.is_stop}")
```

### 3: Extract Only Stopwords

```python
# Define a sentence
text = "The implementation of predictive models requires careful data preprocessing."

doc = nlp(text)

# Initialize empty list
stopwords = []

for token in doc:
  if token.is_stop:
    stopwords.append(token.text)

print("Stopwords :",stopwords)
```

### 4: Extract Non-Stopwords

```python
# sentence
text = "Python provides robust frameworks for efficient database management."
doc = nlp(text)

# Initialize empty texxt
words = []

# Filter out stopwords and punctuation
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.text)
print("Non-Stopwords :",words)
```

### 5: Remove Stopwords from a Sentence

```python
# Sentence
text = "The team is learning basic data science for the new project."
doc = nlp(text)

# Initialize empty list
results = []

# Keep tokens that are not stopwords
for token in doc:
  if not token.is_stop:
    results.append(token.text)

# Join and print the cleaned text
print(" ".join(results))
```

### 6: Remove Stopwords and Punctuation

```python
# Define Sentence
text = "Hello! The manager is checking the data, and the team will help."
doc = nlp(text)
results = []

# Filter out stopwords and punctuation
for token in doc:
  if not token.is_stop and not token.is_punct:
    results.append(token.text)

print(" ".join(results))
```

### 7. Convert Text to Lowercase and Remove Stopwords

```python
# Define Sentence
text = "The Data Team IS Testing the NEW Software Update."

# Process text
doc = nlp(text)

# Initialize empty list
result = []

# Filter stopwords/punctuation and convert to lowercase
for token in doc:
  if not token.is_stop and not token.is_punct:
    result.append(token.text.lower())

# Join and print the lowercase cleaned text
print(" ".join(result))
```

### 9. Add a Custom Stopword

```python
# Add custom stopword
nlp.Defaults.stop_words.add("data")

# Process the new text
doc = nlp("data Science is a very popular field now.")

# Check if each word is a stopword
for token in doc:
  print(f"{token.text:10} -> {token.is_stop}")
```

```python
# Remove the custom stopword
nlp.Defaults.stop_words.remove("data")

doc = nlp("data Science is a very popular field now.")

for token in doc:
  print(token.text, "->", token.is_stop)
```

### 10: Create a Stopword Removal Function

```python
# Define a function to remove stopwords and punctuation
def remove_stopwords(text):

  # Process the input text
  doc = nlp(text)

  # Empty list to store meaningful words
  result = []

  # Loop through each word in the text
  for token in doc:

    # Check if the word is NOT a stopword and NOT a punctuation
    if not token.is_stop and not token.is_punct:
      result.append(token.text)

  # Join the remaining words back into a single string
  return " ".join(result)

# Test the function with a new sentence
text = "The data scientist is building a powerful machine learning model!."

print("Original Text :", text)
print("Cleaned Text :", remove_stopwords(text))
```

### **Stopwords using NLTK**

```python
# Install the Natural Language Toolkit library
!pip install nltk
```

```python
# Import the NLTK library
import nltk

# Download the stopwords dataset from NLTK
nltk.download("stopwords")
nltk.download("punkt")
nltk.download("punkt_tab")

# Import the stopwords module
from nltk.corpus import stopwords

# Load the set of English stopwords innto a variable
stop_words = set(stopwords.words("english"))

# Tokenize the text
from nltk.tokenize import word_tokenize
```

### 1: Display NLTK Stopwords

```python
# Load all English stopwords into a set for faster processing
stop_words = set(stopwords.words("english"))

# print the complete list of English stopwords
print("List of NLTK English Stopwords :\n", stop_words)
```

```python
# Print the total number of stopwords present in the English set
print("Total number of stopwords :", len(stop_words))
```

### 2: Check Whether a word is a Stopword

```python
# Load English stopwords into a set for fast lookup
stop_words = set(stopwords.words("english"))

# Define a new list of words to test
words = ["an", "algorithm", "is", "very", "smart"]

for word in words:
  print(f"{word:12} -> {word.lower() in stop_words}")
```

### 3: Extract Stopwords from a Sentence

```python
# Define the text
text = "Machine learning is a fascinating field for predictive analytics."

# Tokenize the text into individual words
tokens = word_tokenize(text)

# Empty list to store found stopwords
result = []

# Loop through tokens and filter ONLY the stopwords
for word in tokens:
  if word.lower() in stop_words:
    result.append(word)

# Print the extracted stopwords
print("Extracted Stopwords :", result)
```

### 4: Remove Stopwords from a Sentence

```python
# Define a sentence
text = "The developers are building artificial intelligence models."

# Tokenize the sentence into individual words
tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print("Cleaned Tokens :", result)
```

### 5: Remove Stopwords and Punctuation

```python
# Import the string module to access the punctuation list
import string

# Sentence
text = "Wow! The new AI models are predicting weather accurately."

tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words and word not in string.punctuation:
    result.append(word)
print("Cleaned Tokens without Punctuation :", result)
```

### 6: Convert Text to Lowercase and Remove Stopwords

```python
# Define Sentence
text = "Machine Learning Is Transforming The Tech Industry."

tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word.lower())
print("Lowercased Clean Tokens :", result)
```

### 7: Count Stopwords in a Text

```python
# Sentence
text = "Data science is an amazing field for analyzing large datasets."

tokens = word_tokenize(text)

count = 0

for word in tokens:
  if word.lower() in stop_words:
    count += 1
print("Number of stopwords = ", count)
print("Number of non-stopwords = ", len(tokens) - count)
```

### 8: Display Stopwords by Language

```python
# Fetch the list of all available languages in the NLTK stopwords corpus
languages = stopwords.fileids()

# Print the complete list of available languages
print("List of available languages:\n", languages)
```

```python
# Print the exact count of languages available in the NLTK stopwords corpus
print("Total number of languages supported:", len(languages))
```

###  9: Add Custom Stopwords in NLTK

```python
# Define a set of new custom words to be treated as stopwords
custom_words = {"machine", "learning", "technology"}

# Update the existing NLTK stop_words set with our custom words
stop_words.update(custom_words)

text = "Machine learning is an amazing technology today."

# Tokenize the word
tokens = word_tokenize(text)
result = []

for word in tokens:
  if word.lower() not in stop_words:
    result.append(word)
print("Cleaned Tokens :", result)
```
