### `Lemmatization` in NLP using Python

#### **Lemmatization** is the process of converting a word into is base or dictionary form, called a lemma.

## Lemmatization using spaCy

```python
# Install the spaCy library for advanced NLP tasks
!pip install spaCy

# Download the small English language model for spaCy
!python -m spacy download en_core_web_sm
```

```python
# Import spaCy and load the English model:
import spacy
nlp = spacy.load("en_core_web_sm")
```

### 1. Basic Lemmatization

```python
# Define a sentence
text = "The kids are eating sweet apples."

# Process the text to create a spaCy document
doc = nlp(text)

# Loop through each token in the document
for token in doc:
  # Print the original word and its dictionary root word
  print(f"{token.text:10} -> {token.lemma_}")
```

### 2. Lemmatizaing Plural Nouns

```python
# Define text with plural nouns
text = "Trees flowers birds rivers"

doc = nlp(text)

for token in doc:
  print(f"{token.text:10} -> {token.lemma_}")
```

### 3. Lemmatazing Verbs

```python
# words
text = "working worked works"

doc = nlp(text)

for token in doc:
  print(f"{token.text:10} -> {token.lemma_}")
```

### 4. Lemmatizing Different Verb Forms

```python
# Sentence
text = "I am writing. He wrote notes. They have written letters."
doc = nlp(text)
for token in doc:
  if token.is_alpha:
    print(f"{token.text:15} -> {token.lemma_}")
```

### 5. Lemmatizing with a Sentence

```python
text = "The dogs were chasing the fast cats."
doc = nlp(text)
lemmatized_text = " ".join(token.lemma_ for token in doc)
print(lemmatized_text)
```

### 6. Lemmatization of Irregular Words

```python
# Sentence
text = "mice teeth ran ate better"

# Pass the string through the spaCy pipeline
doc = nlp(text)

# Iterate over each word to find its root
for token in doc:

  # Display the original word and its lemma nicely formatted
  print(f"{token.text:10} -> {token.lemma_}")
```

### 7. Lemmatization Based on Part of Speech

```python
# Define the sentence
text = "The fast dogs are chasing mice quickly."

# Analyze the text to get tokens and POS tags
doc = nlp(text)

# Inspect each word, its grammatical role, and its root form
for token in doc:
  print(f"Word : {token.text:10} | POS : {token.pos_:10} | Lemma : {token.lemma_}")
```

### 8. Remove Punctuation and Lemmatize

```python
# Define Sentence
text = "The small cats are jumping, and dogs are eating apples!"

# Process the sentence through tokens
doc = nlp(text)

# Empty list to store the clean root words
words = []

# Loop through all extracted tokens
for token in doc:
  if not token.is_punct:
    words.append(token.lemma_)
print(f"Original Text : {text}")
print(f"Lemmatized Text : {words}")
```

### 9. Lemmatization with Stopword Removal

```python
# sentence
text = "The fast dogs are eating apples, and the cats are jumping!"
doc = nlp(text)
words = []
for token in doc:
  if not token.is_stop and not token.is_punct:
    words.append(token.lemma_)

print(f"Original Text : {text}")
print(f"Filtered list : {words}")
```

### 10. Create a Lemmatization Function

```python
# Define function
def lemmatize_text(text):
  doc = nlp(text)
  result = []
  for token in doc:
    if not token.is_stop and not token.is_punct:
      result.append(token.lemma_)
  return " ".join(result)

text = "The small mice were running quickly, and the cats were eating apples!"
result = lemmatize_text(text)

print(f"Original Text : {text}")
print(f"Final Output : {result}")
```

### Lemmatization using NLTK

#### **NLTK** provides lemmatization primarily through WordNetLemmatize

```python
!pip install nltk
```

```python
# Setup required NLTK data for lemmatization
import nltk

from nltk.corpus import wordnet
```

```python
# Download the WordNet dictionary database
nltk.download("wordnet")

# Download the Open Multilingual WordNet (prevents lookup errors)
nltk.download("omw-1.4")

nltk.download("punkt")
nltk.download("punkt_tab")

nltk.download("averaged_perceptron_tagger")
nltk.download("averaged_perceptron_tagger_eng")
```

```python
# Import the WordNetLemmatizer class from NLTK
from nltk.stem import WordNetLemmatizer

# Create an object instance of the lemmatizer
lemmatizer = WordNetLemmatizer()
```

### 1. Basic NLTK Lemmatization

```python
words = ["cats", "dogs", "cars", "houses"]
for word in words:
  print(f"{word:10} -> {lemmatizer.lemmatize(word)}")
```

### 2. Lemmatizing Verbs

```python
# By default, NLTK assumes a word is a noun.
# Therefore, we should specify the POS when lemmatizing verbs.
words = ["playing", "played", "plays"]
for word in words:
  print(f"{word:10} -> {lemmatizer.lemmatize(word, pos="v")}")
```

### 3. Lemmatinzing Nouns

### 6. Lemmatizing Verbs

```python
word = ["went", "gone", "going", "eaten", "ate"]

for word in words:
  print(f"{word:10} -> {lemmatizer.lemmatize(word, pos = "v")}")
```

### 7. Compare Noun and Verb Lemmatization

```python
# The same word can produce different results depending
word = "meeting"
print("As noun :", lemmatizer.lemmatize(word, pos = "n"))
print("As verb :", lemmatizer.lemmatize(word, pos = "v"))
```

### 8. Lemmatizing a List of Words

```python
words = ["studies",
         "studying",
         "studied",
         "students",
         "books"
]
for word in words:
  if word.endswith("ing") or word.endswith("en"):
    lemma = lemmatizer.lemmatize(word, pos = "v")
  else:
    lemma = lemmatizer.lemmatize(word, pos = "n")

  print(f"{word:10} -> {lemma}")
```

### 9. Lemmatization with Tokenization

```python
from nltk.tokenize import word_tokenize

text = "The students are studying books."

# Tokenize the text
tokens = word_tokenize(text)

for token in tokens:
  lemma = lemmatizer.lemmatize(token)
  print(f"{token:10} -> {lemma}")
```

 ### 10. Complete NLTK Lemmatization with POS

```python
def get_wordnet_pos(word):
  tag = nltk.pos_tag([word])[0][1]

  if tag.startswith("J"):
    return wordnet.ADJ
  elif tag.startswith("V"):
    return wordnet.VERB
  elif tag.startswith("N"):
    return wordnet.NOUN
  elif tag.startswith("R"):
    return wordnet.ADV
  else:
    return wordnet.NOUN

text = "The children are playing with their toys"

tokens = word_tokenize(text)

for word in tokens:
  if word.isalpha():
    pos = get_wordnet_pos(word)
    lemma = lemmatizer.lemmatize(word, pos = pos)
    print(f"{word:10} -> {lemma}")
```
