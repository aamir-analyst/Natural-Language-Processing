## **`Tokenization` in NLP**
**Tokenization :** Computers cannot read large blocks of text at once. Tokenization breaks raw text into smaller, manageable pieces called tokens.
- **Sentence Tokenization :** Cuts the paragraph into individual sentences.

- **Word Tokenization :** Cuts each sentence into individual words and punctuation marks.

### 1. Word Tokenization

```python
# Import the main NLTK library
import nltk

# Import the word tokenize function
from nltk.tokenize import word_tokenize

text = "Python is widely adopted in enterprise software development and data analytics."

# Split the text into individual word tokens
tokens = word_tokenize(text)
print("Tokeniza Text :",tokens)
```

- word_tokenize(text) breaks the sentence into individual tokens.

### 2. Sentence Tokenization

```python
# Import the sentence tokenize functio
from nltk.tokenize import sent_tokenize

# Define a professional multi-sentence text
text = "Natural Language Processing empowers systems to understand human language. The NLTK library provides comprehensive tools for text analysis. Furthermore, Python remains the industry standard for these advanced tasks."

# Split the text into individual sentence
sentences = sent_tokenize(text)

# Output the generated sentence tokens
print("Tokenize Sentence :",sentences)
```

- sent_tokenize(text) is useful when we want to process a document sentence by sentence.

### 3. Word Tokenization of Multiple Sentence

### 8. Sentence Tokenization + Word Tokenization

```python
from nltk.tokenize import sent_tokenize, word_tokenize

# Define multi-sentence paragraph
text = """
Data Science drives modern business strategies.
Python provides robust frameworks for data analysis.
Consequently, organizations rely heavily on these tools to extract insights."""

# First, split the paragraph into individual sentence
sentences = sent_tokenize(text)

# Loop through each extracted sentence
for sentence in sentences:
  words = word_tokenize(sentence)
  print(words)
```

### 9. Count Number of Tokens

```python
# Define a data science statement
text = "Effective data preprocessing is essential, and Python provides powerful libraries for this task."

# Split the text into individual tokens
tokens = word_tokenize(text)

# Output the list of tokens
print("Tokens : ", tokens)

# Output the total count of tokens using the len() function
print("Number of tokens : ", len(tokens))
```

### 10. Count Only Words

```python
from nltk.tokenize import word_tokenize

# Import string module to access the punctuation list
import string

# Define a statement containing punctuation
text = "Data pipelines are essential, and Python offers robust frameworks for data engineering."

# Split the text into tokens
tokens = word_tokenize(text)

# Filter out tokens that are merely punchuation marks using list comprehension
words = [token for token in tokens if token not in string.punctuation]

# Output the original unfiltered tokens
print("All tokens : ", tokens)

# Output the filtered list containing only words
print("Words : ", words)

# Output the exact count of valid words
print("Number of words : ", len(words))
```

### 11. Convert Tokens to Lowercase

```python
# Define a statement with mixed casing
text = "Consistent DATA Normalization is VITAL for Machine Learning."

# Split the text into initial tokens
tokens = word_tokenize(text)

# Convert every token to lowercase using list comprehension
lower_tokens = [word.lower() for word in tokens]

# Output the original tokens to see the mixed cases
print("Original tokens:")
print(tokens)

# Output the normalized lowercase tokens
print("\nLowercase tokens:")
print(lower_tokens)
```

### 12. Tokenization Using wordpunct_tokenize()

```python
# Import the wordunct tokenizer
from nltk.tokenize import wordpunct_tokenize

# Define a statement with punctuation and a contraction
text = "Alert! Server utilization reached 90%. Shouldn't we scale?"

# Tokenize text by splitting it at both whitespaces and punctuation boundaries
tokens = wordpunct_tokenize(text)

# Output the resulting tokens
print("Tokenize Statement :",tokens)
```

### 13. Tokenize a Paragraph

```python
# Define paragraph
text = """
Natural Language Processing (NLP) is a modern branch of artificial intelligence that helps computers understand human language. It bridges the gap between how people communicate and how machines process digital data. Using NLP, systems can read text, translate languages, and analyze emotions from sentences. Every day, tools like virtual assistants, spam filters, and search engines rely heavily on these techniques. In Python, libraries such as regular expressions and text parsers make extracting useful information very simple. As data grows rapidly around the world, learning to handle text data has become an essential skill for every data science student. With regular practice and hands-on projects, mastering text processing becomes straightforward and exciting.
"""
sentences = sent_tokenize(text)
print("Number of sentences :", len(sentences))
for i, sentence in enumerate(sentences, start = 1):
  print(f"\nSentence {i} :")
  print(sentence)

  tokens = word_tokenize(sentence)
  print("Tokens:",tokens)
```

### 15. Complete Tokenization Program

```python
# Define the paragraph
text = """
Python is an industry-leading programming language.
It is extensively utilized across Data Science, Machine Learning, and NLP pipelines!
"""

# Step 1: Sentence Tokenization
sentences = sent_tokenize(text)

print("Sentences:")
for sentence in sentences:
  print(sentence)

# Step 2: Word Tokenization
tokens = word_tokenize(text)

print("\nAll Tokens:")
print(tokens)

# Step 3: Remove punctuation
words = [word for word in tokens if word not in string.punctuation]

print("\nTokens without punctuation:")
print(words)

# Step 4: Convert to lowercase
lower_words = [word.lower() for word in words]

print("\nLowercase tokens:")
print(lower_words)

# Step 5: Count tokens
print("\nTotal tokens:", len(tokens))
print("Total words:", len(words))
```
