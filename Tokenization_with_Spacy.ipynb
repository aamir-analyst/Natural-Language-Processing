### Tokenization Using spacy in python
### Tokenization is the process of breaking text into smaller units called tokens. A token can be a word, punctuation mark, number, or special symbol.

```python
# Then import spacy:

import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
# 2. Basic Tokenization
text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 3. Tokenization with Token Index
# Each token has an index indicating its position in the document.
# token.i gives the position/index fo the token.

import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.i, token.text)
```

```python
# 4. Tokenization with Token Attributes
# spacy provides several useful attributes for each token.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy to learn."
doc = nlp(text)
for token in doc:
  print(token.text,
        token.is_alpha,
        token.is_punct,
        token.is_space)
```

```python
# 10. Tokenization of Hyphenated WOrds
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Data-driven decision-making is important."
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 11. Removing Punctuation After Tokenization
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is powerful, easy, and popular!"
doc = nlp(text)
for token in doc:
  if not token.is_punct:
    print(token.text)
```

```python
# 12. Getting Only Alphabetic Tokens
# We can use is_alpha to select only alphabetic tokens.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python 3 is easy to learn!"
doc = nlp(text)
for token in doc:
    if token.is_alpha:
        print(token.text)
```

```python
# 13. Converting Tokens into a List
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is popular."
doc = nlp(text)
tokens = [token.text for token in doc]
print(tokens)
```

```python
# 14. Tokenization of Multiple Sentences
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy to learn. It is widely used in Data Science."
doc = nlp(text)
for token in doc:
  print(token.text)
```

- Spacy processes the entire text as a Doc object.

```python
# 15. Tokenization and Sentence Segmentation
# spacy can also identify individual sentences.
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy to learn. It is popular in Data Science."
doc = nlp(text)
for sentence in doc.sents:
  print("Sentence:", sentence.text)
  for token in sentence:
    print(token.text)
  print()
```

```python
# 16. Tokenization from a text file
import spacy
nlp = spacy.load("en_core_web_sm")
with open("/content/NLP_text_file.txt", "r", encoding = "utf-8") as file:
  text = file.read()
doc = nlp(text)
for token in doc:
  print(token.text)
```

```python
# 17. Complete Example
import spacy
nlp = spacy.load("en_core_web_sm")
text = "Python is easy to learn! I have 5 years of experience."
doc = nlp(text)
print("Token\tAlpha\tPunctuation")
for token in doc:
  print(token.text,"\t",
        token.is_alpha,"\t",
        token.like_num,"\t",
        token.is_punct)
```
